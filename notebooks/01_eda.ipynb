```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv("../data/cars.csv")

df.head()
```

```python
print("Dimenzije dataseta:", df.shape)

print("\nTipovi podataka:")
print(df.dtypes)

print("\nNedostajuće vrednosti:")
print(df.isnull().sum())
```

```python
print("Statistika numeričkih kolona:")
display(df.describe())
```

```python
print("Broj jedinstvenih vrednosti:")

for column in df.columns:
    print(f"{column}: {df[column].nunique()}")
```

```python
print("Distribucija kategorijskih kolona:\n")

categorical_columns = df.select_dtypes(include="str").columns

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts().head(15))
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df["priceUSD"], bins=50, kde=True)

plt.title("Distribucija cena automobila")
plt.xlabel("Cena (USD)")
plt.ylabel("Broj automobila")

plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df["priceUSD"])

plt.title("Outlieri u ceni automobila")
plt.xlabel("Cena (USD)")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df["year"], bins=30)

plt.title("Distribucija godišta automobila")
plt.xlabel("Godina")
plt.ylabel("Broj automobila")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df["mileage(kilometers)"], bins=50)

plt.title("Distribucija kilometraže")
plt.xlabel("Kilometraža (km)")
plt.ylabel("Broj automobila")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="year",
    y="priceUSD",
    alpha=0.3
)

plt.title("Veza između godišta i cene")
plt.xlabel("Godina")
plt.ylabel("Cena (USD)")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="mileage(kilometers)",
    y="priceUSD",
    alpha=0.3
)

plt.title("Veza između kilometraže i cene")
plt.xlabel("Kilometraža (km)")
plt.ylabel("Cena (USD)")

plt.show()
```
